```python
players = ["messi", "lamin", "ronaldinho"]
goals = [25, 15, 10]
matches = 30
for i,player in enumerate(players): # goals scored by each player
    rate = round(goals[i] /matches,2) # goals by each player per each match
    print(player, rate)
    if rate >0.5:
        print("excellent player")
    else:
        print("Needs improvement")
```

```python
players = ["messi", "lamin", "ronaldinho"]
goals = [25, 15, 10]
matches = 30
for i,player in enumerate(players): # goals scored by each player
    rate = round(goals[i] /matches,2) # goals by each player per each match
    print(player, rate)
    if rate >=0.5: # if the rate more or equal to 0.5 the player level is excellent, unless needs improvement
        print("excellent player")
    else:
        print("Needs improvement")
```
